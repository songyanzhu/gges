- `gdf_cities`: a GeoDataFrame containing several points of London (-0.1278, 51.5074), Bristol (-2.5879, 51.4545), Manchester (-2.2426, 53.4808), Edinburgh (-3.1883, 55.9533).

## Answers

````python
import numpy as np

temperatures = [14, 16, 13, 18, 19]

# Convert the list into a NumPy array
temperatures = np.array(temperatures)

print("Mean:", temperatures.mean())
print("Minimum:", temperatures.min())
print("Maximum:", temperatures.max())
````

````python
import pandas as pd

# Create a table
df = pd.DataFrame({
    "place": ["Highfield", "City Centre", "Southampton Common"],
    "temperature": [16, 18, 15]
})

display(df)

# Select places with temperatures above 15°C
display(df[df["temperature"] > 15])
````

````python
import matplotlib.pyplot as plt

# Plot temperatures at each place
plt.bar(df["place"], df["temperature"], color="orange")
plt.ylabel("Temperature (°C)")
plt.title("Example temperatures in Southampton")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
````

````python
import geopandas as gpd

# Create Southampton as a point using longitude and latitude
gdf_sou = gpd.GeoDataFrame(
    {"city": ["Southampton"]},
    geometry=gpd.points_from_xy([-1.4044], [50.9097]),
    crs="EPSG:4326"
)

# Create points for the other cities
gdf_cities = gpd.GeoDataFrame(
    {"city": ["London", "Bristol", "Manchester", "Edinburgh"]},
    geometry=gpd.points_from_xy(
        [-0.1278, -2.5879, -2.2426, -3.1883],
        [51.5074, 51.4545, 53.4808, 55.9533]
    ),
    crs="EPSG:4326"
)

# Convert both datasets to British National Grid (metres)
gdf_sou = gdf_sou.to_crs(epsg=27700)
gdf_cities = gdf_cities.to_crs(epsg=27700)

# Plot both datasets on the same map
ax = gdf_cities.plot(color="blue", markersize=40, label="Other cities")
gdf_sou.plot(ax=ax, color="red", markersize=60, label="Southampton")
ax.set_xlabel("Easting (m)")
ax.set_ylabel("Northing (m)")
ax.set_title("UK city locations")
ax.legend()
plt.show()

# Calculate the straight-line distance to each city
gdf_cities["distance_m"] = gdf_cities.geometry.distance(
    gdf_sou.geometry.iloc[0]
)

display(gdf_cities[["city", "distance_m"]])

# Find the nearest city
nearest = gdf_cities.loc[gdf_cities["distance_m"].idxmin()]

print("Nearest city:", nearest["city"])
print(f"Shortest distance: {nearest['distance_m']:.1f} metres")
````
